```python
import numpy as np

# 1. 定義適用於批次（2D矩陣）的激發函數與誤差函數
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax_batch(x):
    # 針對每一列（axis=1）進行 softmax 計算，避免數值溢位
    c = np.max(x, axis=1, keepdims=True)
    exp_x = np.exp(x - c)
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

def cross_entropy_error_batch(y_hat, y):
    delta = 1e-7
    batch_size = y_hat.shape[0]
    # 計算整批平均的交叉熵誤差
    return -np.sum(y * np.log(y_hat + delta)) / batch_size

# 2. 產生 5 筆輸入資料 X (形狀: 5x6) 與真實標籤 Y (形狀: 5x3)
np.random.seed(42) # 設定亂數種子以便重現結果
X = np.random.randn(5, 6)
# 假設 5 筆資料各自對應的真實類別分佈 (One-hot 形式)
Y = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0]
])

print("--- 輸入與真實標籤形狀 ---")
print("X 形狀:", X.shape)
print("Y 形狀:", Y.shape)

# 3. 設定各層權重(W)與偏置(b)
# 第一隱藏層: 6 -> 5
W1 = np.random.randn(6, 5)
b1 = np.random.randn(5)

# 第二隱藏層: 5 -> 4
W2 = np.random.randn(5, 4)
b2 = np.random.randn(4)

# 第三隱藏層: 4 -> 4
W3 = np.random.randn(4, 4)
b3 = np.random.randn(4)

# 輸出層: 4 -> 3
W4 = np.random.randn(4, 3)
b4 = np.random.randn(3)

print("\n--- 各層權重與偏值陣列形狀 ---")
print(f"W1: {W1.shape}, b1: {b1.shape}")
print(f"W2: {W2.shape}, b2: {b2.shape}")
print(f"W3: {W3.shape}, b3: {b3.shape}")
print(f"W4: {W4.shape}, b4: {b4.shape}")

# 4. 批次前向傳播計算 (Forward Pass)
z1 = np.dot(X, W1) + b1
h1 = relu(z1)

z2 = np.dot(h1, W2) + b2
h2 = relu(z2)

z3 = np.dot(h2, W3) + b3
h3 = sigmoid(z3)

z4 = np.dot(h3, W4) + b4
y_hat = softmax_batch(z4)

print("\n--- 各層加權和 z 與輸出向量結果與形狀 ---")
print(f"z1 形狀: {z1.shape} | h1 形狀: {h1.shape}")
print(f"z2 形狀: {z2.shape} | h2 形狀: {h2.shape}")
print(f"z3 形狀: {z3.shape} | h3 形狀: {h3.shape}")
print(f"z4 形狀: {z4.shape} | y_hat 形狀: {y_hat.shape}")
print("\ny_hat 輸出結果:\n", y_hat)

# 5. 計算整批 (五筆) y_hat 和 Y 誤差值的平均
loss = cross_entropy_error_batch(y_hat, Y)
print(f"\n整批資料的平均 Cross Entropy Error: {loss:.4f}")
```
